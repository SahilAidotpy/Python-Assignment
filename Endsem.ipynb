```python
# 1.An e-commerce platform tracks user activity:
#  User ID
#  Viewed Products (list)
#  Purchased Products (set)
#  Product Categories (dictionary: product → category)


users = [
    {
        "user_id": 1,
        "viewed": ["p1", "p2", "p3"],
        "purchased": {"p2"}
    },
    {
        "user_id": 2,
        "viewed": ["p1", "p3", "p4"],
        "purchased": {"p4"}
    },
    {
        "user_id": 3,
        "viewed": ["p1", "p3"],
        "purchased": set()
    }
]

product_category = {
    "p1": "Electronics",
    "p2": "Clothing",
    "p3": "Electronics",
    "p4": "Books"
}
```

```python
# (a) Identify products frequently viewed but never purchased
# Step 1: Count viewed products
view_count = {}
purchased_products = set()

for user in users:
    for product in user["viewed"]:
        if product in view_count:
            view_count[product] += 1
        else:
            view_count[product] = 1

    for product in user["purchased"]:
        purchased_products.add(product)

# Step 2: Find products viewed but never purchased
never_purchased = []

for product in view_count:
    if product not in purchased_products:
        never_purchased.append(product)

print("Products viewed but never purchased:", never_purchased)
```

```python
# b. Build a dictionary that maps:

# category → list of users who purchased from that category
# Ensure no duplicate users per category.

category_users = {}

for user in users:
    user_id = user["user_id"]

    for product in user["purchased"]:
        category = product_category[product]

        if category not in category_users:
            category_users[category] = []

        if user_id not in category_users[category]:
            category_users[category].append(user_id)

print("Category to users mapping:", category_users)
```

```python
# (c) Time Complexity Analysis
# N = number of users
# V = average viewed products per user
# P = average purchased products per user

# Complexity for identifying non-purchased viewed products:
# Counting viewed products → O(N * V)
# Collecting purchased products → O(N * P)
# Checking viewed vs purchased → O(N * V)
# O(N(P+V)

```
